## Responses API

```python
from dotenv import load_dotenv
import os
from openai import OpenAI, AsyncOpenAI

load_dotenv(override=True)

if (OPENAI_API_KEY := os.getenv("OPENAI_API_KEY")) is None:
    raise ValueError("OPENAI_API_KEY not found in environment variables")

try:
    llm_client = AsyncOpenAI(api_key=OPENAI_API_KEY)

    system_prompt = "You are a helpful assistant."
    response_stream = await llm_client.responses.create(
        model="gpt-4.1-mini",
        input="Hello, how are you?",
        instructions=system_prompt,
        stream=True,
        store=False
    )

    async for event in response_stream:
        if event and event.type == "response.output_text.delta":
            print(event.delta, end="", flush=True)
except Exception as e:
    print(f"[ERROR] - {str(e)}")
finally:
    await llm_client.close()
```

## Agents SDK

```python
from agents import Agent, OpenAIResponsesModel, Runner

try:
    openai_gpt4_1_mini = OpenAIResponsesModel(
        model="gpt-4.1-mini",
        openai_client=AsyncOpenAI(api_key=OPENAI_API_KEY)
    )

    assistant_agent = Agent(
        name="Assistant",
        instructions=system_prompt,
        model=openai_gpt4_1_mini
    )

    agent_response_stream = Runner.run_streamed(
        starting_agent=assistant_agent,
        input="Hello, how are you?"
    )

    async for event in agent_response_stream.stream_events():
        if event and event.type == "raw_response_event":
            data = event.data
            if data and data.type == "response.output_text.delta":
                print(event.data.delta, end="", flush=True)
except Exception as e:
    print(f"[ERROR] - {str(e)}")
```
